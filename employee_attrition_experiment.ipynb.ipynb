# 01 : using max_depth to minimize tree growth of the model

```python
depth_values = []
training_scores = []
testing_scores = []
#to record class 1 recall and F1 data as well
class1_recalls = []
class1_f1s = []
```

```python
for depth in range(1, 21):

    model_depth = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        random_state=42
    )

    model_depth.fit(X0_train, y0_train)

    train_score = model_depth.score(X0_train, y0_train)
    test_score = model_depth.score(X0_test, y0_test)

    depth_values.append(depth)
    training_scores.append(train_score)
    testing_scores.append(test_score)

    print(
        f"Depth: {depth} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f}"
    )
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import recall_score, f1_score

for depth in range(1, 21):

    model02 = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        random_state=42
    )

    model02.fit(X0_train, y0_train)

    # Accuracy
    train_score = model02.score(X0_train, y0_train)
    test_score = model02.score(X0_test, y0_test)

    # Predictions on TEST data
    preds = model02.predict(X0_test)

    # Specifically evaluate class 1
    recall = recall_score(y0_test, preds)
    f1 = f1_score(y0_test, preds)

    # Store everything
    depth_values.append(depth)
    training_scores.append(train_score)
    testing_scores.append(test_score)

    class1_recalls.append(recall)
    class1_f1s.append(f1)

    print(
        f"Depth: {depth:2d} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f} | "
        f"Recall-1: {recall:.3f} | "
        f"F1-1: {f1:.3f}"
    )
```

```python
plt.plot(depth_values, training_scores, label="Training Accuracy")
plt.plot(depth_values, testing_scores, label="Testing Accuracy")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")
plt.title("Random Forest: max_depth vs Accuracy")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(depth_values, class1_recalls, label="Class 1 Recall")
plt.plot(depth_values, class1_f1s, label="Class 1 F1")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Score")
plt.title("max_depth vs Class 1 Performance")
plt.legend()

plt.show()
```

# 02 : using class_weight + max_depth -> 2 hyperparamter

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import recall_score, f1_score, precision_score

depths = range(1, 21)

train_scores_bal = []
test_scores_bal = []

precision_scores_bal = []
recall_scores_bal = []
f1_scores_bal = []

##start loop for variable tree depth value
for depth in depths:
    ## define model and hyperparameters
    model3 = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        class_weight="balanced",
        random_state=42
    )
    ## fit model
    model3.fit(X0_train, y0_train)
    ## make predictions on test split data
    pred = model3.predict(X0_test)
    ## check scores
    train_score = model3.score(X0_train, y0_train)
    test_score = model3.score(X0_test, y0_test)

    precision = precision_score(y0_test, pred)
    recall = recall_score(y0_test, pred)
    f1 = f1_score(y0_test, pred)

    train_scores_bal.append(train_score)
    test_scores_bal.append(test_score)

    precision_scores_bal.append(precision)
    recall_scores_bal.append(recall)
    f1_scores_bal.append(f1)

    print(
        f"Depth: {depth:2d} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f} | "
        f"Precision-1: {precision:.3f} | "
        f"Recall-1: {recall:.3f} | "
        f"F1-1: {f1:.3f}"
    )
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(depths, train_scores_bal, marker="o",
         label="Training Accuracy")

plt.plot(depths, test_scores_bal, marker="o",
         label="Testing Accuracy")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")
plt.title("Balanced Random Forest: max_depth vs Accuracy")

plt.xticks(depths)
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(depths, precision_scores_bal, marker="o",
         label="Class 1 Precision")

plt.plot(depths, recall_scores_bal, marker="o",
         label="Class 1 Recall")

plt.plot(depths, f1_scores_bal, marker="o",
         label="Class 1 F1")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Score")
plt.title("Balanced Random Forest: max_depth vs Class 1 Performance")

plt.xticks(depths)
plt.ylim(0, 1)
plt.legend()
plt.grid()

plt.show()
```

# 03 min_samples_leaf asks: "How many employees must be sitting in a final leaf before I'm willing to treat that leaf as a meaningful pattern?"

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_score, recall_score, f1_score

leaf_values = range(1, 21)

train_scores_leaf = []
test_scores_leaf = []

precision_scores_leaf = []
recall_scores_leaf = []
f1_scores_leaf = []

for leaf in leaf_values:

    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=7,
        min_samples_leaf=leaf,
        class_weight="balanced",
        random_state=42
    )

    model.fit(X0_train, y0_train)

    pred = model.predict(X0_test)

    train_score = model.score(X0_train, y0_train)
    test_score = model.score(X0_test, y0_test)

    precision = precision_score(y0_test, pred)
    recall = recall_score(y0_test, pred)
    f1 = f1_score(y0_test, pred)

    train_scores_leaf.append(train_score)
    test_scores_leaf.append(test_score)

    precision_scores_leaf.append(precision)
    recall_scores_leaf.append(recall)
    f1_scores_leaf.append(f1)

    print(
        f"Leaf: {leaf:2d} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f} | "
        f"Precision-1: {precision:.3f} | "
        f"Recall-1: {recall:.3f} | "
        f"F1-1: {f1:.3f}"
    )
```

```python
import matplotlib.pyplot as plt

leaf_values = range(1, 21)

plt.figure(figsize=(10,5))

plt.plot(leaf_values, train_scores_leaf,
         marker="o", label="Training Accuracy")

plt.plot(leaf_values, test_scores_leaf,
         marker="o", label="Testing Accuracy")

plt.title("Balanced Random Forest: min_samples_leaf vs Accuracy")
plt.xlabel("Minimum Samples per Leaf")
plt.ylabel("Accuracy")
plt.xticks(leaf_values)
plt.grid(True)
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(leaf_values, precision_scores_leaf,
         marker="o", label="Class 1 Precision")

plt.plot(leaf_values, recall_scores_leaf,
         marker="o", label="Class 1 Recall")

plt.plot(leaf_values, f1_scores_leaf,
         marker="o", label="Class 1 F1")

plt.title("Balanced Random Forest: min_samples_leaf vs Class 1 Performance")
plt.xlabel("Minimum Samples per Leaf")
plt.ylabel("Score")
plt.xticks(leaf_values)
plt.grid(True)
plt.legend()

plt.show()
```

# 05 min_sample_splits : It specifies the minimum number of samples that must be present in a node before the node is allowed to split.

```python
# ==============================
# min_samples_split Experiment
# ==============================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Store results
split_values = []

training_scores_split = []
testing_scores_split = []

precision_scores_split = []
recall_scores_split = []
f1_scores_split = []

# Try different min_samples_split values
for split in range(2, 41, 2):

    model = RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced",
        max_depth=7,
        min_samples_leaf=15,
        min_samples_split=split
    )

    model.fit(X0_train, y0_train)

    pred = model.predict(X0_test)

    train_score = model.score(X0_train, y0_train)
    test_score = model.score(X0_test, y0_test)

    precision = precision_score(y0_test, pred)
    recall = recall_score(y0_test, pred)
    f1 = f1_score(y0_test, pred)

    split_values.append(split)

    training_scores_split.append(train_score)
    testing_scores_split.append(test_score)

    precision_scores_split.append(precision)
    recall_scores_split.append(recall)
    f1_scores_split.append(f1)

    print(
        f"Split: {split:2d} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f} | "
        f"Precision-1: {precision:.3f} | "
        f"Recall-1: {recall:.3f} | "
        f"F1-1: {f1:.3f}"
    )
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(
    split_values,
    training_scores_split,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    split_values,
    testing_scores_split,
    marker="o",
    label="Testing Accuracy"
)

plt.title("Balanced Random Forest: min_samples_split vs Accuracy")
plt.xlabel("Minimum Samples Required to Split")
plt.ylabel("Accuracy")

plt.xticks(split_values)
plt.grid(True)

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    split_values,
    precision_scores_split,
    marker="o",
    label="Class 1 Precision"
)

plt.plot(
    split_values,
    recall_scores_split,
    marker="o",
    label="Class 1 Recall"
)

plt.plot(
    split_values,
    f1_scores_split,
    marker="o",
    label="Class 1 F1"
)

plt.title("Balanced Random Forest: min_samples_split vs Class 1 Performance")
plt.xlabel("Minimum Samples Required to Split")
plt.ylabel("Score")

plt.xticks(split_values)
plt.grid(True)

plt.legend()

plt.show()
```

# 06 using dept. column (oneHotEncoder) + tuned RandomForest 

```python
# Features (keep department)
X_dept = emp.drop(["employee_id", "left_company"], axis=1)

# Labels
y_dept = emp["left_company"]

print(X_dept.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ["department"]

transformer = ColumnTransformer(
    transformers=[
        ("one_hot",
         OneHotEncoder(),
         categorical_features)
    ],
    remainder="passthrough"
)

X_dept = transformer.fit_transform(X_dept)

print(X_dept.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_dept,
    y_dept,
    test_size=0.5,
    random_state=42,
    stratify=y_dept
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model_dept = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced",
    max_depth=7,
    min_samples_leaf=15
)

model_dept.fit(X_train, y_train)
```

```python
pred_dept = model_dept.predict(X_test)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

print("EMPLOYEE ATTRITION MODEL (WITH DEPARTMENT)\n")

print("Training Accuracy:",
      model_dept.score(X_train, y_train))

print("Testing Accuracy:",
      model_dept.score(X_test, y_test))

print("\nClassification Report:\n")
print(classification_report(y_test, pred_dept))

print("Confusion Matrix:\n")
print(confusion_matrix(y_test, pred_dept))
```
