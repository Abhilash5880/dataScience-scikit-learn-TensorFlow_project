```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

emp = pd.read_csv("employee_attrition_synthetic.csv")
```

```python
emp
```

```python
X0=emp.drop(["employee_id", "department", "left_company"], axis=1) 
# no DEPT.
```

```python
#create y(labels)
y0=emp["left_company"]
```

```python
y0
```

```python
model0 = RandomForestClassifier(n_estimators=100, random_state=42)

# We'll keep the default hyperparameters
model0.get_params()
```

```python
X0_train, X0_test, y0_train, y0_test = train_test_split(
    X0,
    y0,
    test_size=0.5,
    random_state=42,
    stratify=y0
)
```

```python
model0.fit(X0_train, y0_train)
```

```python
pred0 = model0.predict(X0_test)
```

```python
pred0
```

```python
print("Training accuracy:",
      model0.score(X0_train, y0_train))

print("Testing accuracy:",
      model0.score(X0_test, y0_test))

print(classification_report(y0_test, pred0))

print(confusion_matrix(y0_test, pred0))
```

# 01 : using max_depth to minimize tree growth of the model

```python
depth_values = []
training_scores = []
testing_scores = []
#to record class 1 recall and F1 data as well
class1_recalls = []
class1_f1s = []
```

```python
for depth in range(1, 21):

    model_depth = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        random_state=42
    )

    model_depth.fit(X0_train, y0_train)

    train_score = model_depth.score(X0_train, y0_train)
    test_score = model_depth.score(X0_test, y0_test)

    depth_values.append(depth)
    training_scores.append(train_score)
    testing_scores.append(test_score)

    print(
        f"Depth: {depth} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f}"
    )
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import recall_score, f1_score

for depth in range(1, 21):

    model02 = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        random_state=42
    )

    model02.fit(X0_train, y0_train)

    # Accuracy
    train_score = model02.score(X0_train, y0_train)
    test_score = model02.score(X0_test, y0_test)

    # Predictions on TEST data
    preds = model02.predict(X0_test)

    # Specifically evaluate class 1
    recall = recall_score(y0_test, preds)
    f1 = f1_score(y0_test, preds)

    # Store everything
    depth_values.append(depth)
    training_scores.append(train_score)
    testing_scores.append(test_score)

    class1_recalls.append(recall)
    class1_f1s.append(f1)

    print(
        f"Depth: {depth:2d} | "
        f"Train: {train_score:.3f} | "
        f"Test: {test_score:.3f} | "
        f"Recall-1: {recall:.3f} | "
        f"F1-1: {f1:.3f}"
    )
```

```python
plt.plot(depth_values, training_scores, label="Training Accuracy")
plt.plot(depth_values, testing_scores, label="Testing Accuracy")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")
plt.title("Random Forest: max_depth vs Accuracy")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(depth_values, class1_recalls, label="Class 1 Recall")
plt.plot(depth_values, class1_f1s, label="Class 1 F1")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Score")
plt.title("max_depth vs Class 1 Performance")
plt.legend()

plt.show()
```
